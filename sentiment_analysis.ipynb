```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "sentiment-analysis"
)
```

```python
result = classifier(
    "Beautiful"
)

print(result)
```

```python
result = classifier(
    "I hate it"
)

print(result)
```

```python
result = classifier(
    "Thhis movie is okay"
)

print(result)
```

```python
while True:

  text = input("Enter text:")

  if text.lower()== "exit":
    break

  result=classifier(text)

  print(result)
```
